```python
import pandas as pd

# 1. Загрузка английского датасета
print("Загружаем английский датасет...")
df_en = pd.read_csv('../data/train.csv')
toxic_columns = ['toxic', 'severe_toxic', 'obscene', 'threat', 'insult', 'identity_hate']
df_en['is_toxic'] = df_en[toxic_columns].sum(axis=1).apply(lambda x: 1 if x > 0 else 0)
df_en = df_en[['comment_text', 'is_toxic']]
df_en = df_en.rename(columns={'comment_text': 'text', 'is_toxic': 'toxic'})
df_en['lang'] = 'en'

# 2. Загрузка русского датасета
print("Загружаем русский датасет...")
df_ru = pd.read_csv('../data/labeled.csv')
if 'comment' in df_ru.columns:
    df_ru = df_ru.rename(columns={'comment': 'text'})
df_ru['lang'] = 'ru'
df_ru = df_ru[['text', 'toxic', 'lang']]

# 2.5 Загрузка КАЗАХСКОГО датасета
print("Загружаем казахский датасет...")
# Укажи тут правильное название твоего файла
df_kk = pd.read_csv('../data/toxic_comments.csv', on_bad_lines='skip')

# Приводим к стандарту: переименовываем comment в text
if 'comment' in df_kk.columns:
    df_kk = df_kk.rename(columns={'comment': 'text'})

# Оставляем только нужные колонки
df_kk = df_kk[['text', 'toxic', 'lang']]


# 3. Объединение ТРЕХ датасетов
print("Склеиваем данные (En + Ru + Kk)...")
df_merged = pd.concat([df_en, df_ru, df_kk], ignore_index=True)


# 4. Перемешивание (Shuffle)
print("Перемешиваем строки...")
df_merged = df_merged.sample(frac=1, random_state=42).reset_index(drop=True)


# 5. Сохранение результата
print("Сохраняем итоговый файл...")
df_merged.to_csv('merged_toxicity_dataset.csv', index=False)

print("\n--- ГОТОВО ---")
print(f"Общий размер нового датасета: {len(df_merged)} строк.")
print("Распределение языков:\n", df_merged['lang'].value_counts())
print("\nПервые 5 строк для проверки:")
print(df_merged.head())
```
